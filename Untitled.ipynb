```python
import numpy as np
import pandas as pd

df = pd.read_csv('https://docs.google.com/spreadsheets/d/e/2PACX-1vSGLlRGkcE6OZwelTRA0vVfUvd0328oY8x57td6VyqTtmwr1kLn8vWFZ77VpO4YBDtGlRl5usWMn5qL/pub?gid=1297221839&single=true&output=csv', na_values='NaN')
df.head()
```

```python
df.dtypes
```

```python
df['Stars'] = df['Stars'].astype('int32')
df['True Stars'] = df['True Stars'].astype('int32')
df['Destruction'] = df['Destruction'].astype('int32')

df = df[(df['Defender TH'] >= 18)]
df
```

```python
avg_stars = df.groupby("Attacker Tag")["Stars"].mean().round(2)
avg_stars
```

```python
stats = (df.groupby(["Attacker", "Attacker Tag"])
        .agg(
          Average_Stars=("Stars", "mean"),
          Count=("Stars", "count"))
        .reset_index())
stats = stats.sort_values(
    ["Count", "Average_Stars"],
    ascending=[False, False]
)
pd.set_option("display.max_rows", None)
display(stats)
```
